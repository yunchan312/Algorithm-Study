```python
class FixedSizeStack:
    def __init__(self, size):
        self.stack = []
        self.max_size = size
        self.size = 0
        self.top = -1

    def __str__(self):
        if self.size == 0:
            return "empty"
        return (f"stack: {self.stack}")

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == self.max_size

    def push(self, data):
        if self.is_full():
            print("full")
            return
        self.stack.append(data)
        self.size +=1
        self.top += 1

    def Pop(self):
        if self.is_empty():
            print("empty")
            return
        self.size -= 1
        self.top -= 1
        return self.stack.pop()

    def peek(self):
        if self.is_empty():
            print("empty")
            return
        return self.stack[self.top]

    def Size(self):
        return self.size
```

```python
class ListQueue:
    def __init__(self):
        self.item = []
    def __str__(self):
        if len(self.item) == 0:
            return "empty"
        return f"{self.item}"

    def is_empty(self):
        return len(self.item) == 0

    def size(self):
        return len(self.item)

    def dequeue(self):
        if self.is_empty():
            print("empty")
            return
        return self.item.pop(0)

    def enqueue(self, data):
        self.item.append(data)

    def peek(self):
        if self.is_empty():
            print("empty")
            return
        return self.item[-1]
```

```python
class CircularQueue:
    def __init__(self, max_size):
        self.max_size = max_size
        self.item = [None] * max_size
        self.head = 0
        self.tail = 0
        self.size = 0

    def __str__(self):
        if self.size == 0:
            return "empty"
        res = "Head -> "
        idx = self.head
        for _ in range(self.size):
            res += str(self.item[idx]) + " -> "
            idx = (idx + 1) % self.max_size
        return res + "Tail"

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == self.max_size

    def enqueue(self, data):
        if self.is_full():
            print("full")
            return
        self.item[self.tail] = data
        self.tail = (self.tail + 1) % self.max_size
        self.size += 1

    def dequeue(self):
        if self.is_empty():
            print("empty")
            return
        t = self.item[self.head]
        self.item[self.head] = None
        self.head = (self.head+1)%self.max_size
        self.size -= 1
        return t
    def peek(self):
        return self.item[self.head]
```

```python
if __name__ == "__main__":
    cq = CircularQueue(5)

    print("== 초기 상태 ==")
    print(cq)  # empty

    print("\n== enqueue 3개 추가 ==")
    cq.enqueue(10)
    cq.enqueue(20)
    cq.enqueue(30)
    print(cq)  # Head -> 10 -> 20 -> 30 -> Tail

    print("\n== peek 테스트 ==")
    print("peek:", cq.peek())  # 10

    print("\n== dequeue 2개 ==")
    print("dequeue:", cq.dequeue())  # 10
    print("dequeue:", cq.dequeue())  # 20
    print(cq)  # Head -> 30 -> Tail

    print("\n== enqueue 3개 추가해서 wraparound 테스트 ==")
    cq.enqueue(40)
    cq.enqueue(50)
    cq.enqueue(60)  # 이제 tail이 head를 한 바퀴 돌아서 차야 함
    print(cq)  # Head -> 30 -> 40 -> 50 -> 60 -> Tail

    print("\n== 꽉 찬 상태에서 enqueue ==")
    cq.enqueue(70)  # full 출력

    print("\n== dequeue 전부 ==")
    while not cq.is_empty():
        print("dequeue:", cq.dequeue())
    print(cq)  # empty

    print("\n== 다시 enqueue 테스트 ==")
    cq.enqueue(100)
    cq.enqueue(200)
    print(cq)  # Head -> 100 -> 200 -> Tail

    print("\n== is_empty, is_full 테스트 ==")
    print("is_empty:", cq.is_empty())  # False
    print("is_full:", cq.is_full())    # False
```
