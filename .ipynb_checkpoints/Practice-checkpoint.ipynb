```python
poem_BobDylan = '''
Mama, take this badge off of me.
I can't use it anymore.
It's gettin' dark, too dark to see.
I feel I'm knockin' on Heaven's door.
knock knock knockin' on Heaven's door.
Mama, put my guns in the ground.
I can't shoot them anymore.
That long black cloud is comin' down.
I feel I'm knockin' on Heaven's door.
knock knock knockin' on Heaven's door.
'''

def countChar_dictionary(poem):
    poem_dict = {} 
    poem = poem.lower()
    for p in poem:
        if p.isalpha():
            if p in poem_dict.keys():
                poem_dict[p] += 1
            else:
                poem_dict[p] = 1

    max = 0
    idx = ''
    for k, v in poem_dict.items():
        if v > max:
            max = v
            idx = k
    poem_dict['most'] = idx
    return poem_dict

countChar_dictionary(poem_BobDylan)
```

```python
import math

class LinearList:
    def __init__(self):
        self.item = []

    def __str__(self):
        return f"{self.item}"

    def Append(self, data):
        l = self.Size() + 1
        new_list = [None] * l
        for k,v in enumerate(self.item):
            new_list[k] = v
        new_list[l-1] = data
        self.item = new_list

    def RaiseError(self, index):
        if index < 0 or index > self.Size():
            raise Exception("Index Out of Range")
    
    # def Add(self, index, data):
    #     self.RaiseError(index)
    #     if index == self.Size():
    #         self.item.append(data)
    #     else:
    #         upper = self.item[:index]
    #         lower = self.item[index:]
    #         upper.append(data)
    #         upper.extend(lower)
    #         self.item = upper

    def Add(self, index, data):
        self.RaiseError(index)
        if index == self.Size():
            self.Append(data)
        else:
            new = [None] * (len(self.item) + 1)
            for i in range(index):
                new[i] = self.item[i]
            new[index] = data
            for j in range(index+1, (len(self.item)+1)):
                new[j] = self.item[j-1]

    def Size(self):
        return len(self.item)

    def Delete(self, index):
        self.RaiseError(index)
        target = self.item[index]
        upper = self.item[:index]
        lower = self.item[index+1:]
        upper.extend(lower)
        self.item = upper
        return target

    def Search(self, data):
        for k, d in enumerate(self.item):
            if d == data:
                return k
        print("No Data")
        return 

    def Count(self, data):
        count = 0
        for d in self.item:
            if d == data:
                count += 1
        return count

    def Is_empty(self):
        return self.Size() == 0

    def Min(self):
        if self.Is_empty():
            raise Exception("list is Empty")
        min = math.inf
        for d in self.item:
            if min > d:
                min = d
        return min

    def Max(self):
        if self.Is_empty():
            raise Exception("list is Empty")
        max = self.Min()
        for d in self.item:
            if max < d:
                max = d
        return max
            
        
```

```python
l = LinearList()
l.Add(0,0)
print(l)
l.Add(0,1)
print(l)
l.Add(0,2)
l.Add(0,3)
print(l)
print(l.Delete(0))
print(l)
l.Search(10)
l.Count(0)
l.Append(100)
min = l.Min()
max = l.Max()
print(l)
print(f"max: {max} | min: {min}")
```
