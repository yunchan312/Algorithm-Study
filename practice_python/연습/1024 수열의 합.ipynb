```python
N, L = map(int, input().split())

def soa(l):
    if l == 1:
        return 1
    else:
        return (l*(l+1))/2

arrayLength = L
rest = soa(arrayLength-1)

while True:
    if (N-rest)%arrayLength == 0:
        if N-rest < 0 or arrayLength > 100:
            print(-1)
            break
        start = ((N-rest)//arrayLength)
        for i in range(arrayLength):
            print(int(start + i*1), end=" ")
        break
    else:
        rest += arrayLength
        arrayLength += 1
```

1. arrayLength > 100일 때 -1이라는 조건 놓침

```python
N, L = map(int, input().split())

def getStart(N, L):
    return (N - (L * (L - 1) // 2)) / L

while True:
    s = getStart(N, L)
    if s < 0 or L > 100:
        print(-1)
        break
    if int(s) == s:
        for i in range(L):
            print(int(s+i), end=" ")
        break
    else:
        L += 1
```
