```python
from collections import deque

N = int(input())
target = list(enumerate(map(int, input().split())))

idxList = deque([])
P = deque([])
for i in range(N-1, -1, -1):
    for j in range(i):
        if target[j+1][1] < target[j][1]:
            target[j+1], target[j] = target[j], target[j+1]
    idxList.appendleft(target[i][0])
    
idxList = (list(enumerate(idxList)))

for i in range(N-1, -1, -1):
    for j in range(i):
        if idxList[j+1][1] < idxList[j][1]:
            idxList[j+1], idxList[j] = idxList[j], idxList[j+1]
    P.appendleft(idxList[i][0])
    
for p in list(P):
    print(p, end=" ")
```
