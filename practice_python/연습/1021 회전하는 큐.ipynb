```python
from collections import deque

N, M = map(int, input().split())
targets = list(map(int, input().split()))
# N, M = 32, 6
# targets = [27, 16, 30, 11, 6, 23]
circularQueue = deque([i+1 for i in range(N)])

headIdx = 0
ans = 0
for target in targets:
    target = target
    targetIdx = circularQueue.index(target)
    diff = abs(headIdx - targetIdx)
    lengDiff = abs(len(circularQueue) - diff)
    headIdx = targetIdx
    if diff <= lengDiff:
        ans += diff
    else:
        ans += lengDiff
    circularQueue.remove(target)
print(ans)
```
