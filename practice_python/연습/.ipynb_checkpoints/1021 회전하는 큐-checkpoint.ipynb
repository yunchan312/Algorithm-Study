```python
N, M = map(int, input().split())
targets = list(map(int, input().split()))
circularQueue = [i+1 for i in range(M)]

ans = 0
head = 0
for i in targets:
    target = i - 1
    print(f"target: {target}")
    toLeft = abs(head - target)
    toRight = abs(len(circularQueue) - target + 1)
    if toLeft <= toRight:
        print(f"head: {head} | currentQueue: {circularQueue}")
        ans += toLeft
        head = (head + toLeft)%len(circularQueue)
        circularQueue.pop(head)
        head = (head+1)%len(circularQueue)
    else:
        print(f"head: {head} | currentQueue: {circularQueue}")
        ans += toRight
        head = (abs(head - toRight))%len(circularQueue)
        circularQueue.pop(head)
        head = (head+1)%len(circularQueue)
        


print(ans)
        

```
