```python
times = int(input())

for _ in range(times):
    N, M = map(int, input().split())
    dp = [[m+1 for m in range(M)] for _ in range(N)]
    if N == 1:
        print(M)
    else:
        for n in range(1, N):
            for m in range(n, M):
                if n == m:
                    dp[n][m] = 1
                else:
                    dp[n][m] = dp[n][m-1] + dp[n-1][m-1]
        print(dp[N-1][M-1])
```
