```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def __str__(self):
        current = self.head
        result = []
        while current:
            result.append(str(current.data))
            current = current.next
        return " -> ".join(result)
        
    def append(self, data):
        if self.size == 0:
            self.head = Node(data)
            self.tail = self.head
        else:
            prevTail = self.tail
            new_node = Node(data)
            self.tail.next = new_node
            self.tail = new_node
        self.size += 1

    def pop(self, index):
        if self.size == 0:
            raise Exception("Empty")
        if index < 0 or index >= self.size:
            raise Exception("Index out of RANGE!!!!")
        
        if index == 0:
            targetNode = self.head
            self.head = self.head.next
            if self.size == 1:
                self.tail = None
        else:
            prevNode = self.head
            for _ in range(index - 1):
                prevNode = prevNode.next
            targetNode = prevNode.next
            prevNode.next = targetNode.next
            if index == self.size - 1:
                self.tail = prevNode
        self.size -= 1
        return targetNode.data
        
# ---------------------------------------
# times = int(input())
times = 1


for _ in range(times):
    M, N, C = map(int, input().split()) # M이 가로 N이 세로
    field = [[0 for _ in range(M)] for _ in range(N)]
    cabbages = [(0, 0), (1, 0), (1, 1), (4, 2), (4, 3), (4, 5), (2, 4), (3, 4), (7, 4), (8, 4), (9, 4), (7, 5), (8, 5), (9, 5), (7, 6), (8, 6), (9, 6)]
    
    # for _ in range(C):
    #     x, y = map(int, input().split())
    #     field[y][x] = 1
    #     cabbages.append((x, y))

    for y, x in cabbages:
        field[x][y] = 1

    stack = []
    count = 0
    while len(cabbages) > 0:
        current = cabbages.pop(0)
        stack.append(current)
        if field[current[1]][current[0]] == 1:
            field[current[1]][current[0]] = 0
    
            #상 검사
            if current[1] + 1 < N and field[current[1]+1][current[0]] == 1:
                print(f"{current}: top!")
                print(f"stack: {stack}(<- top)")
                stack.append((current[0],current[1]+1))
                field[current[1]+1][current[0]] = 0
            # under
            elif current[1] - 1 > 0 and field[current[1]-1][current[0]] == 1:
                print(f"{current}: under!")
                print(f"stack: {stack}(<- under)")
                stack.append((current[0], current[1]-1))
                field[current[1]-1][current[0]] = 0
            #left
            elif current[0] - 1 > 0 and field[current[1]][current[0] - 1] == 1:
                print(f"{current}: left!")
                print(f"stack: {stack}(<- left)")
                stack.append((current[0]-1, current[1]))
                field[current[1]][current[0]-1] = 0
            #right
            elif current[0] + 1 < M and field[current[1]][current[0] + 1] == 1:
                print(f"{current}: right!")
                print(f"stack: {stack}(<- right)")
                stack.append((current[0]+1, current[1]))
                field[current[1]][current[0]+1] = 0
            else:
                stack.pop()
                if len(stack) == 0:
                    count += 1
    print(count)
            
            
```

```python
from collections import deque

times = int(input())
for _ in range(times):
    M, N, C = map(int, input().split()) # M이 가로 N이 세로
    cabbages = deque([])
    for _ in range(C):
        x, y = map(int, input().split())
        cabbages.append((x, y))
        
    if C == 1:
        print(1)
    else:
        
        stack = [cabbages.popleft()]
        count = 0
        
        while cabbages:
            if len(stack) > 0:
                x, y = stack[-1]
                # upper
                if (x, y+1) in cabbages and y + 1 < N:
                    stack.append((x, y+1))
                    cabbages.remove((x, y+1))
                elif (x, y-1) in cabbages and y-1 >= 0:
                    stack.append((x, y-1))
                    cabbages.remove((x, y-1))
                elif (x-1, y) in cabbages and x-1 >= 0:
                    stack.append((x-1, y))
                    cabbages.remove((x-1, y))
                elif (x+1, y) in cabbages and x+1 < M:
                    stack.append((x+1, y))
                    cabbages.remove((x+1, y))
                else:
                    stack.pop()
                    
            else:
                count += 1
                stack.append(cabbages.popleft())
        if stack:
            count += 1
            
        print(f"{count}")
```
